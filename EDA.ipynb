**Download and connect to the database file containing survey data.**


To start, download and load the dataset into a `pandas` DataFrame.



```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)
# Set pandas option to display all columns
pd.set_option('display.max_columns', None)
## Replace missing values
df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum()
df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])
# Display the first few rows of the dataset
df.head()
```

### Analysis of Experience and Job Satisfaction

```python
## Experience range binning for 'YearsCodePro'
df['YearsCodePro'] = df['YearsCodePro'].replace({'Less than 1 year': 0.5, 'More than 50 years': 51})
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

clean_df = df[['YearsCodePro', 'JobSat']].dropna()

bins = [0, 5, 10, 20, 51]
labels = ['0-5 yrs', '5-10 yrs', '10-20 yrs', '>20 yrs']

clean_df['ExperienceRange'] = pd.cut(clean_df['YearsCodePro'], bins=bins, labels=labels, include_lowest=True)

## Find average Job Satisfaction Score for each experience range
median_sat = clean_df.groupby('ExperienceRange')['JobSat'].median().reset_index()

plt.figure(figsize=(10, 6))
ax = sns.barplot(data=median_sat, x='ExperienceRange', y='JobSat', palette='viridis')
plt.title('Median Job Satisfaction by Coding Experience Range')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction (0–10)')
plt.ylim(0, 10)
for container in ax.containers:
    ax.bar_label(container, fmt='%.f', label_type='edge', padding=-20, color='white')
plt.tight_layout()
plt.show()
```

```python
## Count of distribution for average Job Satisfaction
plt.figure(figsize=(10,6))
sns.countplot(x='JobSat', data=df, palette='viridis')
plt.xlabel('Job Satistfaction Score (0-10)')
plt.ylabel('Count')
plt.title('Distribution of Job Satisfaction Scores')
plt.show()
```

### Programming Languages Comaparison between Have Worked With and Want to Work With


```python
## Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`
from collections import Counter

# Split the semicolon-separated strings into lists
have_worked = df['LanguageHaveWorkedWith'].dropna().str.split(';')
want_to_work = df['LanguageWantToWorkWith'].dropna().str.split(';')

# Flatten the lists and count frequencies
have_counts = Counter([lang.strip() for sublist in have_worked for lang in sublist])
want_counts = Counter([lang.strip() for sublist in want_to_work for lang in sublist])

# Convert counters to DataFrames
have_df = pd.DataFrame(have_counts.items(), columns=['Language', 'HaveWorkedWith'])
want_df = pd.DataFrame(want_counts.items(), columns = ['Language', 'WantToWorkWith'])

# Alternative
# have_df = pd.DataFrame.from_dict(have_counts, orient='index', columns=['HaveWorkedWith'])
# want_df = pd.DataFrame.from_dict(want_counts, orient='index', columns=['WantToWorkWith'])

# Merge both counts on language
combined_df = pd.merge(have_df, want_df, on='Language', how='outer').fillna(0)

# Sort by total popularity
combined_df['Total'] = combined_df['HaveWorkedWith'] + combined_df['WantToWorkWith']
combined_df = combined_df.sort_values(by='Total', ascending=False).drop(columns='Total')
combined_df

# Limit to top 10 for clarity
top_n = 15
top_combined = combined_df.head(top_n)

# Plot
top_combined.plot(kind='bar', x='Language', figsize=(10,8), color=['#6B5B95', '#88B04B'])
plt.title('Top Programming Languages: Have Worked With vs Want To Work With')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.legend(title='Language Category')
plt.tight_layout()
plt.show()
```

### Analyzing Compensation Outliers

```python
## Calculate basic statistics and outliers for over 3 Standard Deviations above the mean for ConvertedCompYearly
mean_comp = df['ConvertedCompYearly'].mean()
median_comp = df['ConvertedCompYearly'].median()
std_comp = df['ConvertedCompYearly'].std()
print(f'ConvertedCompYearly: mean {mean_comp}, median {median_comp}, std {std_comp}')

threshold = mean_comp + 3 * std_comp
print(threshold)
outliers_comp = df[df['ConvertedCompYearly'] > threshold]
# outliers_comp['ConvertedCompYearly'].min()
outliers_comp
```

```python
## Plot using IQR method
Q1 = df['ConvertedCompYearly'].quantile(0.25)
Q3 = df['ConvertedCompYearly'].quantile(0.75)
IQR = Q3 - Q1
print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")

outliers_iqr = df[
    (df['ConvertedCompYearly'] < lower_bound) | 
    (df['ConvertedCompYearly'] > upper_bound)
]
print(f"Number of outliers detected: {outliers_iqr.shape[0]}")

plt.figure(figsize=(12, 6))
sns.boxplot(x=df['ConvertedCompYearly'], palette='viridis')
plt.title('Box Plot of ConvertedCompYearly (with Outliers)')
plt.xlabel('Yearly Compensation')
plt.xlim(0, 1000000)
plt.show()
```
