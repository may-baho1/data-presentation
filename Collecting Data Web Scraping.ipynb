# **Lab Web Scraping**


## Extract compensation information 

```python
#this url contains the salary data to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The relevant data is the *name of the programming language* and *average annual salary*.<br> 

```python
# Import the beautifulsoup libraries
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
# Download the webpage
r=requests.get(url)
data = r.text
```

```python
# Create soup 
soup = BeautifulSoup(data, 'html.parser')
```

```python
# Scrape the `language name` and `annual average salary' information
table = soup.find('table')
language_data = []
for i, row in enumerate(table.find_all('tr')):
    if i == 0:
        continue  # skip the header row
    cols = row.find_all('td')
    name = cols[1].text.strip()
    avg_sal = cols[3].text.strip()
    entry = {"Name":name, "Average Salary": avg_sal}
    language_data.append(entry)
df = pd.DataFrame(language_data)
df
```

Saving the scrapped data into a file named *popular-languages.csv*

```python
# Save the scrapped data into a file named *popular-languages.csv*
df.to_csv("popular-languages.csv", index=False)
```
